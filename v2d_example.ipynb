```python
# Import libraries
import v2d
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# Set parameters for line vector example
VECTOR_PATH_01 = 'https://github.com/francobarrionuevoenv21/raster_distance_py/raw/refs/heads/main/data_test/rutas_prov_rep03.geojson'
PX_M_01 = 10
OUTPUT_01 = 'routes_map'

# Set parameters for polygon vector example
VECTOR_PATH_02 = 'https://github.com/francobarrionuevoenv21/raster_distance_py/raw/refs/heads/main/data_test/areas_urbanas_rep.geojson'
PX_M_02 = 30
OUTPUT_02 = 'urb_dist_map'
```

```python
# Create distance maps and export results
dist_map_01 = v2d.vector_to_distance(VECTOR_PATH_01, PX_M_01, OUTPUT_01)
dist_map_02 = v2d.vector_to_distance(VECTOR_PATH_02, PX_M_02, OUTPUT_02)
```

```python
#Plot results
fig, ax = plt.subplots(2, 2, figsize=(13, 5), tight_layout=True)
ax = ax.flatten()

# Line vector and distance map
gpd.read_file(VECTOR_PATH_01).plot(ax=ax[0], color='black')
ax[0].set_title('Line Vector (Target: Provincial Routes)', fontsize=13)
ax[0].set_xticks([])
ax[0].set_yticks([])

im1 = ax[1].imshow(dist_map_01)
fig.colorbar(im1, label= 'Distance to target [m]')
ax[1].set_title(f'Distance Map to Target ({PX_M_01} m)', fontsize=13)
ax[1].set_xticks([])
ax[1].set_yticks([])

# Polygon vector and distance map
gpd.read_file(VECTOR_PATH_02).plot(ax=ax[2], color='red')
ax[2].set_title('Polygon Vector (Target: Urban Areas)', fontsize=13)
ax[2].set_xticks([])
ax[2].set_yticks([])

im2 = ax[3].imshow(dist_map_02)
fig.colorbar(im2, label= 'Distance to target [m]')
ax[3].set_title(f'Distance Map to Target ({PX_M_02} m)', fontsize=13)
ax[3].set_xticks([])
ax[3].set_yticks([])
```
